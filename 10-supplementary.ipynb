# Supplementary material

## Load test dataset

```python
import xarray as xr

## Open a netCDF file in a xarray dataset
fname = 'data/garachico2048.ens.nc'
ds = xr.open_dataset(fname)
da = ds['tephra_col_mass']
```

## Plot configuration

```python
import matplotlib.pyplot as plt
import cartopy.crs as crs                        # coordinate systems for maps
import cartopy.feature as cfeature

plt.rcParams['figure.dpi'] = 200

BORDERS = cfeature.NaturalEarthFeature(
        scale     = '10m',
        category  = 'cultural',
        name      = 'admin_0_countries',
        edgecolor = 'gray',
        facecolor = 'none'
        )
LAND = cfeature.NaturalEarthFeature(
        'physical', 'land', '10m',
        edgecolor = 'none',
        facecolor = 'lightgrey',
        alpha     = 0.8
        )

conf = dict(cmap='hot', levels=[5,10,15,20,25,30,35,40], extend='max')
```

```python
## Points for histograms

coords = [[65, 42],[52, 40],[70, 54],[53, 51]]
```

## Plot ensemble mean of test dataset

```python
fig, ax = plt.subplots( subplot_kw={'projection': crs.PlateCarree()} )

cs = ax.contourf(da.lon,da.lat,da.mean(dim='ens'),**conf)

cbar = fig.colorbar(cs,
                    shrink = 0.8,
                    #pad = 0.05,
                    #orientation = 'horizontal',
                    label = r'Volcanic ash column mass [$g~m^{-2}$]')

ax.set_title('Ensemble mean (test dataset)')
ax.set_extent([-20, -11, 23, 30]) # [x1,x2,y1,y2]
ax.add_feature(LAND,zorder=0)
ax.add_feature(BORDERS, linewidth=0.4)
###
### Add grid lines
###
gl = ax.gridlines(
    crs         = crs.PlateCarree(),
    draw_labels = ['left','bottom'],
    linewidth   = 0.5,
    color       = 'gray',
    alpha       = 0.5,
    linestyle   = '--')
gl.xlabel_style  = {'size': 8}
gl.ylabel_style  = {'rotation': 89, 'size': 8}

###
### Points positions
###
for i, coord in enumerate(coords):
    ix,iy = coord
    lon, lat = da[:,iy,ix].lon.item(),da[:,iy,ix].lat.item()
    print(f"point {i}: {lon}, {lat}")
    va = 'bottom'
    ha = 'right'
    xshift, yshift = 0,0
    if i==3:
        va = 'top'
        ha = 'center'
        yshift = -0.05
    ax.plot(lon,lat,'g+',ms=4,alpha=0.8)
    ax.text(lon+xshift,lat+yshift,f"P{i+1}",
            ha = ha,
            va = va,
            fontsize=7
           )
```
